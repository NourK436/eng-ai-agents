$$p(o \mid c) = \frac{\exp(\mathbf{c}\cdot\mathbf{v}_o)}{\sum_{w}\exp(\mathbf{c}\cdot\mathbf{v}_w)},
\qquad L = -\log p(o \mid c).$$

```python
# solution
import plotly.graph_objects as go

def plot_embeddings(coords, words, query=None, neighbor_set=None):
    neighbor_set = set(neighbor_set) if neighbor_set else set()
    idx_query = [i for i, w in enumerate(words) if w == query]
    idx_nbrs  = [i for i, w in enumerate(words) if w in neighbor_set and w != query]
    idx_bg    = [i for i, w in enumerate(words) if w != query and w not in neighbor_set]

    def scatter(idx, name, color, size, opacity=0.8):
        if not idx:
            return None
        return go.Scatter3d(
            x=coords[idx, 0], y=coords[idx, 1], z=coords[idx, 2],
            mode='markers', name=name,
            text=[words[i] for i in idx],
            hovertemplate='%{text}<extra></extra>',
            marker=dict(size=size, color=color, opacity=opacity)
        )

    traces = [t for t in [
        scatter(idx_bg,    'words',     '#4488cc', 3, 0.5),
        scatter(idx_nbrs,  'neighbors', '#ff8800', 6, 0.9),
        scatter(idx_query, 'query',     '#ff2222', 9, 1.0),
    ] if t is not None]

    fig = go.Figure(data=traces)
    fig.update_layout(
        title='Embedding Projector' + (f' — query: "{query}"' if query else ''),
        scene=dict(xaxis_title='PC1', yaxis_title='PC2', zaxis_title='PC3'),
        margin=dict(l=0, r=0, b=0, t=40)
    )
    return fig

plot_embeddings(pca3, plot_words).show()
```

```python
# solution
def neighbors(word, k=10):
    idx   = word2idx[word]
    vec   = emb[idx]
    norms = np.linalg.norm(emb, axis=1)
    sims  = emb.dot(vec) / (norms * np.linalg.norm(vec) + 1e-10)
    ranked = np.argsort(-sims)
    result = []
    for i in ranked:
        if i == idx:
            continue
        result.append((idx2word[i], float(sims[i])))
        if len(result) == k:
            break
    return result

for w, s in neighbors('government', 10):
    print(f"{w:15s} {s:.3f}")
```

### Question 1: Neighborhood quality across word types

Testing a few different words reveals some clear patterns in how well the embeddings capture meaning.

**Nouns** tend to return the cleanest neighbors. Querying "government" brings up words like parliament, administration, senate, and congress — all clearly in the same semantic field. Similarly, "army" clusters with "military", "troops", "forces", and "soldiers". The model has clearly learned something real about topical similarity here.

**Verbs** are more mixed. A concrete verb like "attack" still returns fairly coherent neighbors ("assault", "raid", "strike"), but a more general verb like "said" ends up near other common reporting verbs ("noted", "stated", "told"), which is semantically reasonable even if less impressive.

**Function words** (e.g., "the", "of", "in") return neighbors that are other function words. That's technically correct — they co-occur in similar syntactic environments — but it's not very informative since these words don't carry much content.

**Why rare words give noisier neighbors:** Word2Vec learns by observing co-occurrence patterns across many examples. A frequent word like "government" appears thousands of times across many different contexts, which gives the optimizer plenty of signal to push its embedding into a stable, well-defined position in the space. A rare word might only appear a handful of times, so its embedding is shaped by just a few co-occurrences — which could be coincidental or unrepresentative. The gradient updates are also less frequent, meaning the embedding never fully converges. The result is a vector that ends up somewhere in the space more by chance than by meaning.
